```python
arquivo_1 = origem + 'EXP_2025.csv'
arquivo_2 = origem + 'EXP_2026.csv'
ncm = origem + 'NCM.csv'
pais = origem + 'PAIS.csv'
vias = origem + 'VIA.csv'
urf = origem + 'URF.csv'
```

```python
exp25 = pd.read_csv(arquivo_1,low_memory=False,sep=';',encoding='utf-8')
exp26 = pd.read_csv(arquivo_2,low_memory=False,sep=';',encoding='utf-8')
expncm = pd.read_csv(ncm,low_memory=False,sep=';',encoding='latin-1')
exppais = pd.read_csv(pais,low_memory=False,sep=';',encoding='latin-1')
expvias = pd.read_csv(vias,low_memory=False,sep=';',encoding='utf-8')
expurf = pd.read_csv(urf,low_memory=False,sep=';',encoding='latin-1')
```

```python
exp_final = pd.concat([exp25,exp26], ignore_index=True)
```

```python
exp_final = exp_final.merge(exppais[['CO_PAIS','NO_PAIS']],on='CO_PAIS',how='left')
exp_final = exp_final.merge(expncm[['CO_NCM','NO_NCM_POR']],on='CO_NCM',how='left')
```

```python
exp_final.to_csv(origem+'exp_final.csv',index=False)
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Seus dados já estão carregados em 'exp_final'

# 1. Filtrar os dados de janeiro de 2025 a janeiro de 2026
exp_filtrado = exp_final[
    ((exp_final['CO_ANO'] == 2025) & (exp_final['CO_MES'] >= 1)) |
    ((exp_final['CO_ANO'] == 2026) & (exp_final['CO_MES'] <= 1))
]

# 2. Criar uma coluna combinada para agrupamento (ANO-MÊS)
exp_filtrado['ANO_MES'] = exp_filtrado['CO_ANO'].astype(str) + '-' + exp_filtrado['CO_MES'].astype(str).str.zfill(2)

# 3. Agrupar por mês e somar o KG_LIQUIDO
exp_mensal = exp_filtrado.groupby('ANO_MES')['KG_LIQUIDO'].sum().reset_index()

# 4. Ordenar por data (janeiro de 2025 a janeiro de 2026)
# Vamos criar uma ordem correta
meses_ordem = ['2025-01', '2025-02', '2025-03', '2025-04', '2025-05', '2025-06',
               '2025-07', '2025-08', '2025-09', '2025-10', '2025-11', '2025-12',
               '2026-01']

# Filtrar apenas os meses que existem nos dados
meses_existentes = [mes for mes in meses_ordem if mes in exp_mensal['ANO_MES'].values]
exp_mensal = exp_mensal[exp_mensal['ANO_MES'].isin(meses_existentes)]

# Reordenar
exp_mensal['ANO_MES'] = pd.Categorical(exp_mensal['ANO_MES'], categories=meses_existentes, ordered=True)
exp_mensal = exp_mensal.sort_values('ANO_MES')

# 5. Criar o gráfico
plt.figure(figsize=(14, 7))

# Criar barras com cores gradientes ou mais atrativas
bars = plt.bar(exp_mensal['ANO_MES'], exp_mensal['KG_LIQUIDO'],
               color='steelblue', edgecolor='navy', alpha=0.8)

# Adicionar título e labels
plt.title('Total de Importação em Quilogramas - Janeiro de 2025 a Janeiro de 2026',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Mês', fontsize=13, fontweight='bold')
plt.ylabel('Total em Quilogramas (KG)', fontsize=13, fontweight='bold')

# Formatar o eixo Y para mostrar números grandes de forma legível
plt.ticklabel_format(style='plain', axis='y')
plt.grid(True, axis='y', alpha=0.3, linestyle='--')

# Adicionar os valores sobre as barras
for i, (bar, valor) in enumerate(zip(bars, exp_mensal['KG_LIQUIDO'])):
    # Formatar o valor com separadores de milhar
    valor_formatado = f'{valor:,.0f}'.replace(',', '.')
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + (bar.get_height() * 0.01),
             valor_formatado, ha='center', va='bottom', fontsize=10, fontweight='bold')

# Ajustar o layout
plt.xticks(rotation=45, ha='right')
plt.tight_layout()

# Mostrar o gráfico
plt.show()

# 6. Salvar o gráfico (opcional)
plt.savefig(origem + 'importacao_mensal_jan2025_jan2026.png', dpi=300, bbox_inches='tight')

# 7. Exibir os dados resumidos
print("\n" + "="*60)
print("RESUMO MENSAL DAS IMPORTAÇÕES EM QUILOGRAMAS")
print("="*60)
print(exp_mensal.to_string(index=False))
print("="*60)
print(f"TOTAL GERAL: {exp_mensal['KG_LIQUIDO'].sum():,.0f} KG".replace(',', '.'))
print("="*60)
```
